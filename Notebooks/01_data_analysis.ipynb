```python
import yfinance as yf
```

```python
tickers = ['AAPL', 'MSFT', 'JPM', 'XOM', 'GLD']
```

```python
print(tickers)
```

```python
prices=yf.download(tickers, start='2021-01-01', end='2026-01-01', auto_adjust=True)
```

```python
prices.columns
```

```python
prices.head()
```

```python
prices['Close'].head()
```

```python
close_prices = prices['Close']
```

```python
close_prices.head()
```

```python
print(close_prices)
```

```python
close_prices.info
```

```python
close_prices.describe()
```

```python
close_prices.isna().sum()
```

```python
import matplotlib.pyplot as plt 
close_prices.plot(figsize=(12, 6))
plt.title('Asset Closing Prices')
plt.xlabel('Date')
plt.ylabel('Price')
plt.show()
```

```python
returns = close_prices.pct_change()
returns.head()
```

```python
returns=returns.dropna()
returns.describe()
```

```python
returns.hist(figsize=(12, 8), bins=50)
plt.suptitle('Distribution of Daily Returns')
plt.show()
```

```python
volatility = returns.std()
volatility.plot(kind='bar', figsize=(12, 6))
plt.title('Daily Volatility of Assets')
plt.xlabel('Assets')
plt.ylabel('Volatility')
plt.show()
```

```python
annual_volatility = volatility * (252 ** 0.5)
annual_volatility.plot(kind='bar', figsize=(12, 6))
plt.title('Annualized Volatility of Assets')
plt.xlabel('Assets')
plt.ylabel('Annualized Volatility')
plt.show()
```

```python
cumulative_returns=(1 + returns).cumprod()
cumulative_returns.plot(figsize=(12, 6))
plt.title('Cumulative Returns of Assets')
plt.xlabel('Date')
plt.ylabel('Cumulative Returns')
plt.show()
```

##Observations
- The assets showed different levels of volatility daily and annually.
- GLD had the lowest volatility daily and annually.
- AAPL had the highest volatility daily and annually.
- The equity assets showed more volatility in general in comparison to GLD.
- AAPL, GLD and MSFT showed relatively similar cumulative performance over the same period, finishing at broadly comparable levels.
